```python
#Encuentra el número de reseñas de usuarios y la calificación promedio para cada libro.
query_7 = '''
SELECT
    ratings.book_id AS book_id,
    COUNT(reviews.review_id) AS review_id_cnt
FROM
    ratings
    INNER JOIN reviews ON reviews.book_id = ratings.book_id
GROUP BY
     ratings.book_id
ORDER BY
     ratings.book_id DESC;
'''
```

```python
#Encuentra el número de reseñas de usuarios y la calificación promedio para cada libro.
query_8 = '''
SELECT
    ratings.book_id AS book_id,
    AVG(ratings.rating) AS rating_avg
FROM
    ratings
    INNER JOIN reviews ON reviews.book_id = ratings.book_id
GROUP BY
    ratings.book_id
ORDER BY
    ratings.book_id DESC;
'''
```

```python

#Identifica la editorial que ha publicado el mayor número de libros con más de 50 páginas.
query_9 = '''

SELECT
    publishers.publisher AS publisher_name,
    COUNT(books.publisher_id) AS publisher_cnt
FROM
    books
    INNER JOIN publishers ON books.publisher_id = publishers.publisher_id
WHERE
    num_pages > 50
GROUP BY
    publishers.publisher
ORDER BY
    publisher_cnt DESC;
'''
```

```python
pd.io.sql.read_sql(query_9, con = engine)
```

```python
#El editorial que ha publicado el mayor número de libros con más de 50 páginas es Penguin Books.
```

```python
#Identifica al autor que tiene la más alta calificación promedio del libro: mira solo los libros con al menos 50 calificaciones.
query_10 = '''
SELECT
    books.title AS title,
    authors.author AS author,
    COUNT(ratings.book_id) AS book_id_cnt,
    MAX(ratings.rating) AS rating_max
FROM
    books
    INNER JOIN authors ON authors.author_id = books.author_id
    INNER JOIN ratings ON ratings.book_id = books.book_id
GROUP BY
     books.title, authors.author
HAVING
    COUNT(ratings.book_id) >= 50
ORDER BY
    book_id_cnt DESC, rating_max DESC;
'''
```

```python
pd.io.sql.read_sql(query_10, con = engine)
```

```python
#El autor que tiene la más alta calificación promedio del libro es Stephenie Meyer.
```

```python
#Encuentra el número promedio de reseñas de texto entre los usuarios que calificaron más de 50 libros.
query_11 = '''
SELECT
    username,
    ((COUNT(username) * 100 / 160)::real) AS username_avg,
    SUM(book_id) AS book_id_sum
FROM
    reviews
GROUP BY
     username
HAVING
    SUM(book_id) >= 50
ORDER BY
    username_avg DESC;
'''
```

```python
pd.io.sql.read_sql(query_11, con = engine)
```
